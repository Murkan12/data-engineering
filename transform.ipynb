```python
new_rows = []
for index, row in data_filtered.iterrows():
    if not pd.isna(row['Utworzono zapis w Fenku']):
        new_rows.append((row['ID zapisu'], 'Utworzono zapis w Fenku', row['Utworzono zapis w Fenku']))
    if not pd.isna(row['Data zapisu w LangLionie']):
        new_rows.append((row['ID zapisu'], 'Data zapisu w LangLionie', row['Data zapisu w LangLionie']))
    if not pd.isna(row['Przekazano do DOK']):
        new_rows.append((row['ID zapisu'], 'Przekazano do DOK', row['Przekazano do DOK']))
    if not pd.isna(row['Kliknięto przesłano umowę']):
        new_rows.append((row['ID zapisu'], 'Kliknięto przesłano umowę', row['Kliknięto przesłano umowę']))
    if not pd.isna(row['Nauczyciel potwierdził realizację']):
        new_rows.append((row['ID zapisu'], 'Nauczyciel potwierdził realizację', row['Nauczyciel potwierdził realizację']))
```

```python
activities_df = pd.DataFrame(new_rows, columns=['ID Zapisu', 'Aktywność', 'Data'])
```
